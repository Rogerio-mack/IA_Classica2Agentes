```python
from unified_planning.shortcuts import *

# Remover créditos
get_environment().credits_stream = None

# =========================
# Problema
# =========================

problem = Problem("robot_box")

# =========================
# Fluentes
# =========================

robot_at_a = Fluent("robot_at_a")
robot_at_b = Fluent("robot_at_b")

box_at_a = Fluent("box_at_a")
box_at_b = Fluent("box_at_b")

carrying = Fluent("carrying")
hands_free = Fluent("hands_free")

# =========================
# Adicionar fluentes
# =========================

problem.add_fluent(robot_at_a, default_initial_value=False)
problem.add_fluent(robot_at_b, default_initial_value=False)

problem.add_fluent(box_at_a, default_initial_value=False)
problem.add_fluent(box_at_b, default_initial_value=False)

problem.add_fluent(carrying, default_initial_value=False)
problem.add_fluent(hands_free, default_initial_value=False)

# =========================
# Estado inicial
# =========================

problem.set_initial_value(robot_at_a, True)
problem.set_initial_value(box_at_a, True)

problem.set_initial_value(hands_free, True)

# =========================
# move_ab
# =========================

move_ab = InstantaneousAction("move_ab")

move_ab.add_precondition(robot_at_a)

move_ab.add_effect(robot_at_a, False)
move_ab.add_effect(robot_at_b, True)

# =========================
# move_ba
# =========================

move_ba = InstantaneousAction("move_ba")

move_ba.add_precondition(robot_at_b)

move_ba.add_effect(robot_at_b, False)
move_ba.add_effect(robot_at_a, True)

# =========================
# pickup
# =========================

pickup = InstantaneousAction("pickup")

pickup.add_precondition(robot_at_a)
pickup.add_precondition(box_at_a)
pickup.add_precondition(hands_free)

pickup.add_effect(carrying, True)
pickup.add_effect(hands_free, False)

pickup.add_effect(box_at_a, False)

# =========================
# drop
# =========================

drop = InstantaneousAction("drop")

drop.add_precondition(robot_at_b)
drop.add_precondition(carrying)

drop.add_effect(carrying, False)
drop.add_effect(hands_free, True)

drop.add_effect(box_at_b, True)

# =========================
# Adicionar ações
# =========================

problem.add_action(move_ab)
problem.add_action(move_ba)

problem.add_action(pickup)
problem.add_action(drop)

# =========================
# Objetivo
# =========================

problem.add_goal(box_at_b)

# =========================
# Resolver
# =========================

with OneshotPlanner(name="pyperplan") as planner:

    result = planner.solve(problem)

# =========================
# Exibir plano
# =========================

print("\nPlano encontrado:\n")

if result.plan is not None:

    for i, action in enumerate(result.plan.actions):

        print(f"{i+1}. {action}")

else:
    print("Nenhum plano encontrado.")
```
